```python
import numpy as np
import matplotlib.pyplot as plt
import sys
import os

cwd = os.getcwd()
root_path = os.path.abspath(os.path.join(cwd, os.pardir))
sys.path.append(root_path)
```

```python
from implementation.inverse_power_method import inverse_power_method

A = np.array([[4, 1], [2, 3]])

# Find the eigenvalue of smallest magnitude (shift=0 targets the eigenvalue closest to 0)
eigenvalue, eigenvector = inverse_power_method(A, shift=0.0)

print("Matrix A:")
print(A)
print(f"\nSmallest-magnitude eigenvalue (inverse power method): {eigenvalue:.6f}")
print(f"Corresponding eigenvector: {eigenvector}")

# Compare with numpy
np_eigenvalues = np.real_if_close(np.linalg.eigvals(A))
print(f"\nNumPy eigenvalues: {np.sort(np_eigenvalues)}")
print(
    f"Smallest-magnitude eigenvalue matches: {np.isclose(eigenvalue, np_eigenvalues[np.argmin(np.abs(np_eigenvalues))])}"
)
```

```python
# Visualize convergence by running iterations manually
n = A.shape[0]
x = np.ones(n)
x = x / np.linalg.norm(x)
eigenvalue_estimates = []

for i in range(50):
    y = np.linalg.solve(A, x)
    y = y / np.linalg.norm(y)
    est = np.dot(y, A @ y)
    eigenvalue_estimates.append(est)
    x = y

plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(eigenvalue_estimates) + 1),
    eigenvalue_estimates,
    marker="o",
    markersize=4,
)
plt.axhline(
    y=np_eigenvalues[np.argmin(np.abs(np_eigenvalues))],
    color="r",
    linestyle="--",
    label=f"True smallest-magnitude λ = {np_eigenvalues[np.argmin(np.abs(np_eigenvalues))]:.4f}",
)
plt.title("Convergence of Inverse Power Method")
plt.xlabel("Iteration")
plt.ylabel("Eigenvalue Estimate")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
